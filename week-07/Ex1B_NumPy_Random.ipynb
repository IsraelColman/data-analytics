## Generating Random Numbers

### Random set of integers with 2 elements

```python
r_1 = np.random.randint(1, 20, size=(2))

print(r_1)
```

### Random set of integers with 1 row and 2 columns

```python
r_2 = np.random.randint(1, 20, size=(1, 2))

print(r_2)
```

### Random set of floats with 2 row and 1 columns

```python
r_3 = np.random.rand(2, 1)

print(r_3)
```

### Random set of integers with 4 row and 5 columns

```python
r_4 = np.random.randint(20, 100, size=(4, 5))

print(r_4)
```

### Random set of floats with 4 row and 5 columns

```python
r_5 = np.random.rand(4, 5)

print(r_5)
```

## Generating Random Choices from an Array

### Creating a 1d array with student names from my class

```python
students = np.array([
    "Maream", "Alizon", "Earl", "Faiz",
    "Arash", "Abdul", "Stephanie", "Netsa",
    "Ceasar", "Henry", "Ezekiel", "Meric"])
```

### Calling on one random student from the array

```python
call_on = np.random.choice(students)

print(call_on)
```

### Calling the array to create 3 groups of 4 to be assinged as breakout groups

```python
breakout_groups = np.random.choice(students, size=(3, 4))

print(breakout_groups) 
# This would not be an effective way to assign break out rooms.
# From what i observed, there are repeat names in the array. 
# 1 person cannot be in the same group nor can they be in another group.
```

### Calling with a probability

```python
call_on_weighted = np.random.choice([
    "Maream", "Alizon", "Earl", "Faiz",
    "Arash", "Abdul", "Stephanie", "Netsa",
    "Ceasar", "Henry", "Ezekiel", "Meric"],
    p=[0.15, 0.10, 0.10, 0.30, 0.07, 0.10, 0.04, 0.05, 0.05, 0.02, 0.02, 0.00], size=(3, 4)
)

print(call_on_weighted)

# P values:
# Maream - 0.15 
# Alizon - 0.10 
# Earl - 0.10 
# Faiz - 0.30 
# Arash - 0.07 
# Abdul - 0.10 
# Stephanie - 0.04 
# Netsa - 0.05 
# Ceasar - 0.05 
# Henry - 0.02 
# Ezekiel - 0.02 
# Meric - 0.00
```

## Generating a Normal Distribution

### Using Bell curve

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))

print(bell_curve)
```
