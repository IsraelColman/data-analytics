# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

### Creating the data frame

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
    }

df_sales = pd.DataFrame(data)
df_sales
```

### Adding revenue to data frame with values (units sold * price)

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

### Calculating total units sold and total revenue

```python
total_units_sold = np.sum(df_sales['Units_Sold'])
total_revenue = np.sum(df_sales['Revenue'])
print(f'Total units sold: {total_units_sold}')
print(f'Total revenue: ${total_revenue:,}')
```

### Calculating Average units sold and Average revenue

```python
avg_units_sold = np.mean(df_sales['Units_Sold'])
avg_revenue = np.mean(df_sales['Revenue'])
print(f'Average units sold: {avg_units_sold:,.2f}')
print(f'Average revenue: ${avg_revenue:,.2f}')
```

## Grouping Data

### Grouping product and calculating sum of units sold and revenue based on products

```python
product_totals = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
product_totals
```

### Grouping region and calculating avg of units sold and revenue based on region

```python
Region_avg = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
Region_avg
```

### Grouping product and region. calculating sum of units sold and revenue based on products and region

```python
prod_reg_totals = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()
prod_reg_totals
```

## Sorting and Ranking Data

### Sorting units sold in ascending order

```python
units_sold_sorted = df_sales.sort_values('Units_Sold', ascending=True)
units_sold_sorted
```

### Sorting revenue in descending order

```python
revenue_sorted = df_sales.sort_values('Revenue', ascending=False)
revenue_sorted
```

### Adding new column 'Revenue_Rank' to df_sales and ranking revenue as descending

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False)
```

## Merging DataFrames

### Creating new data frame 'df_products'

```python
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']}

df_products = pd.DataFrame(product_data)
```

### Merging data frames 'df_sales' and 'df_products'

```python
df_merged = pd.merge(df_sales, df_products)
```

### Calling data frame 'df_merged'

```python
df_merged
```

### Calculating the sum of revenue by category

```python
cat_total_revenue = df_merged.groupby('Category')[['Revenue']].sum()
cat_total_revenue
```
