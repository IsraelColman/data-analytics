# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

### Creating x and y values then creating line graph

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(
    x,
    y
)
plt.show()
```

### Updating line graph to include a marker

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(
    x,
    y,
    marker='o'

)
```

### Updating line graph by adding labels

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    marker='o'
)
```

## Styling Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='o'
)
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='s',
    ms=10,
    mfc='green'
)
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='*',
    ms=10,
    mfc='orange'
)
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

### Creating a bar graph

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(
    products,
    sales
)
```

### Adding labels, bar color, bar width to chart

```python
plt.title("Top Performing Product's")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.bar(
    products,
    sales,
    color='skyblue',
    width=0.5
)
```

### Creating a scatter chart

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(
    x,
    y
)
```

### Adding marker size and color map

```python
size = np.random.randint(1, 100, 50)

plt.scatter(
    x,
    y,
    s=size,
    cmap='viridis'
)
```

## Subplots

### Creating a figure with two subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
```

### Plotting a sine wave

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

### Creating plot

```python
axes[0].set_title("Sine wave")
axes[0].set_xlabel("X")
axes[0].set_ylabel("sin(x)")
axes[0].plot(x, y)
```

### Creating histogram

```python
random_numbers = np.random.randn(1000)

axes[1].set_title("Histogram of Random Numbers")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")
axes[1].hist(random_numbers);
```

### calling it all (The chart didn't show before so i wrote everything in one cell block)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

x = np.linspace(0, 10, 100)
y = np.sin(x)

axes[0].plot(x, y)
axes[0].set_title("Sine Wave")
axes[0].set_xlabel("X")
axes[0].set_ylabel("sin(x)")

random_numbers = np.random.randn(1000)

axes[1].hist(random_numbers)
axes[1].set_title("Histogram of Random Numbers")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
