# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

### Creating x and y values then creating line graph

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(
    x,
    y
)
plt.show()
```

### Updating line graph to include a marker

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(
    x,
    y,
    marker='o'

)
```

### Updating line graph by adding labels

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    marker='o'
)
```

## Styling Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='o'
)
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='s',
    ms=10,
    mfc='green'
)
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.plot(
    x,
    y,
    ls='--',
    c='red',
    marker='*',
    ms=10,
    mfc='orange'
)
```
