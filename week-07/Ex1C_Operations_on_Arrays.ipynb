## Permutations

### Print the current value of r_4

```python
print(r_4)
```

### Shuffle the values of r_4

```python
np.random.shuffle(r_4)
print(r_4)
```

each row stayed the same however they were shuffled by placement. for example the row on axis 0 is now a row on axis 2.

### Shuffle with premutation

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print()
print(r_4)
```

I would use shuffle if i know i don't need the original array anymore. If i still need the original array but want another array with the same values
then i would use the permutation.

### Adding array([1])

```python
print(np.add(r_4, np.array([1])))

# the value 1 is added to each element of r_4
```

### Adding array([[1]])

```python
print(np.add(r_4, np.array([[1]])))

# I got the same result as before. the value 1 was added to each element of r_4
```

### Adding array([1, 1, 1, 1, 1])

```python
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))

# same result, value 1 was added to each column
```

### Adding array([[1], [1]])

```python
print(np.add(r_4, np.array([[1], [1]])))

# I get a value error similar to the one before. 
# operands could not be broadcast together with shapes (4,5) (2,1) 
```

Broadcasting is how NumPy lets you do math with arrays that have different shapes, as long as the shapes are compatible. When using np.add(), NumPy matches up the values in each array and adds them together. If one array only has a single value or a dimension of size 1, NumPy repeats that value as needed to match the other array. The shape of the result is usually based on the larger array after the broadcasting rules are applied.

## Grocery Problem

### Creating arrays of number of groceries

```python
bag_1 = np.array([2, 3, 4, 0])

bag_2 = np.array([0, 4, 2, 1])

print(bag_1)
print(bag_2)
```

### Printing how many total items in each bag

```python
print(f'The first bag contains {np.sum(bag_1)} items, and the second bag contains {np.sum(bag_2)} items')
```

### Printing how many vegetables, fruits, boxes of snacks, and beverages there are between both bags

```python
print(f'There are {np.add(bag_1[0], bag_2[1])} vegetables, {np.add(bag_1[1], bag_2[0])} fruits, {np.add(bag_1[2], bag_2[3])} boxes of snacks, and {np.add(bag_1[3], bag_2[2])} beverages in the two grocery bags')
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])
```

```python
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array(
    [
        [ali],
        [ben],
        [carla]
    ]
)

print(all_knocks)
```

```python
print(f'Ali total knocks: {np.sum(all_knocks[0])}\nBen total knocks: {np.sum(all_knocks[1])}\nCarla total knocks: {np.sum(all_knocks[2])}')
print(f'Total Knocks: {np.sum(all_knocks)}')
```

```python
daily_totals = np.sum(all_knocks, axis=0)
rolling_totals = np.cumsum(daily_totals)
estimated_votes = rolling_totals / 2

print(f'Daily totals: {daily_totals}')
print(f'\nRolling totals: {rolling_totals}')
print(f'\nEstimated votes: {estimated_votes}')
```
