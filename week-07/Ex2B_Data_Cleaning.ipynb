# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd
```

### Importing test data

```python
df = pd.read_csv("test_data.csv")
df
```

### Displaying statistics

```python
print(df.describe())
```

## Filtering a DataFrame

### Filtering age

```python
age_filter = df[df['Age'] > 18]
print(age_filter)
```

### Filtering score

```python
score_filter = df[(df['Score'] > 50) & (df['Score'] < 100)]
print(score_filter)
```

## Modifying a DataFrame

### Removing all scores that are less than 0

```python
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

### Removing column 'City'

```python
df_cleaned = df_cleaned.drop(columns='City')
df_cleaned
```

### Replacing na values in age with the mean of age

```python
df_cleaned = df_cleaned.fillna({'Age': df_cleaned['Age'].mean()})
df_cleaned
```

### Replacing the any 0 score with mean of score 

```python
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
```

### Removing any other nan rows in data frame

```python
df_cleaned = df_cleaned.dropna()
df_cleaned
```

### Setting the index keys to the 'ID' column. In turn removes 'ID' column

```python
df_cleaned = df_cleaned.set_index('ID')
df_cleaned
```
