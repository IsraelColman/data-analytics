## Looping Through Arrays

### Creating 2d array

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])
```

### Looping through 1st dimension of arrA

```python
for x in arrA:
    print(x * 2)
```

### Looping through 2nd dimension of arrA

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

### Looping through the 1st dimension gives me the rows multiplied. Looping through the 2nd dimension give me the individual numbers multiplied.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result) 

# it prints out the rows * 2. but since y has 3 numbers, the loop runs 3 times for each row.
```

## Joining and Stacking Arrays

### Concatenating arrA with variable joined_array_1

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

### Concatenating arrA with variable joined_array_2

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

### Stacking joined array 1

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

### Stacking joined array 2

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# The error is telling me that the arrays must have the same shape in order to be stacked.
```

### Another concatenated array

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

### Another concatenated array

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# the arrays that are being concatenated must share the same size. Joined array has more rows than arrA
```

```python
stack_1 = np.hstack((arrA, arrA))
print(stack_1) # np.hstack joins the arrays horizontally

print()

stack_2 = np.vstack((arrA, arrA))
print(stack_2) # np.vstack joins the arrays vertically

print()

stack_3 = np.dstack((arrA, arrA))
print(stack_3) # np.dstack joins the arrays and turns them into a 3 dimensional array.
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky) # it gave me 2 arrays and both are 2 dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino) 
# i get 3 arrays however since it isnt able to split them evenly 3 ways, i get one 2 dimensional array and two 1 dimensional arrays
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([
    [7, 6],
    [13, 5],
    [7, 3]
])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```
