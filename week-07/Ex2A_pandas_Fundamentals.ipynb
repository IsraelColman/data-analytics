## Series

### Creating Collections of values

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {
    "first": 13,
    "second": 14,
    "third": 15,
    "fourth": 16
}
```

### Creating series from the collections

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac)) # Pandas no longer creates Series directly from sets because sets are unordered.
ser4 = pd.Series(ad)

print(ser1)
print(ser2)
print(ser3)
print(ser4)
```

## Coding a DataFrame

### Creating a dictionary

```python
test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Domininc', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'Sand Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]

}
```

### Turning the dictionary into a data frame

```python
df4 = pd.DataFrame(test_data)
df4
```

### Exporting the data frame into a csv file

```python
df4.to_csv("test_data.csv", index=False)
```
