# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    return part_code.split(':')[0]
```

```python
def get_product_num(part_code):
    return part_code.split(':')[1].split('-')[0]
```

```python
def get_size(part_code):
    return part_code.split('-')[1]
```

```python
part1 = 'ACME:123-L'
```

```python
part2 = 'DI:12345-M'
```

```python
part3 = 'ACE:1-12'
```

```python
print(f'{get_supplier_code(part1)}')
print(f'{get_product_num(part1)}')
print(f'{get_size(part1)}')
```

```python
print(f'{get_supplier_code(part2)}')
print(f'{get_product_num(part2)}')
print(f'{get_size(part2)}')
```

```python
print(f'{get_supplier_code(part3)}')
print(f'{get_product_num(part3)}')
print(f'{get_size(part3)}')
```
