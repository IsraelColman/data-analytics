```python
# Can you add a float to an integer? What happens? 
int_example = 10
float_example = 10.50

total = int_example + float_example

print(type(int_example))
print(type(float_example))

print(total, type(total))

# Yes you can, python automatically turns the output into a float.
```

```python
# Can you add a string to an integer? What happens?
str_example = 'Israel'
int_example = 25

israel_age = str_example + int_example

type(str_example)
type(int_example)

print(israel_age, type(israel_age))
```

```python
# Can you add a boolean to an integer? What happens?
boolean_example = True
int_example = 15

boolean_int = boolean_example + int_example

print(type(boolean_example))
print(type(int_example))

print(boolean_int, type(boolean_int))

# yes you can, the boolean gets converted into an integer of 1
```

```python
# Can you add a boolean to a string? What happens?
boolean_example = True
str_example = 'Israel'

boolean_str = boolean_example + str_example

print(type(boolean_example))
print(type(str_example))

print(boolean_str, type(boolean_str))
```

```python
# What happens when you divide an integer by an integer? 
int_1 = 20
int_2 = 50

divided = int_1 / int_2
print(divided)

# no problems at all. the division goes through
```

Write an expression to illustrate the answer to each question below. Make sure to run the cell to check your results.

```python
# What is the result of 2 times 2?

num_1 = 2
num_2 = 2

print(num_1 * num_2)
```

```python
# What is the result of 4 divided by 2, using regular division?
num_1 = 4
num_2 = 2

print(num_1 / num_2)
```

```python
# What is the result of 4 divided by 2, using floor division?
num_1 = 4
num_2 = 2

print(num_1 // num_2)
```
