## Calculating social security tax, medicare tax, and federal tax of person 1, 2, and 3

### Social Security Tax Function

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
    
```

### Medicare Tax Function

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

### Federal Tax Function

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else:
        extra = withholding_code - 4
        tax_rate = 0.18 - (extra * 0.005)

    return gross_pay * tax_rate
```

### Person 1

```python
# Social Security Tax
ss_tax = get_soc_sec_tax(750)
print(f'Social Security Tax: ${ss_tax:.2f}')

# Medicare Tax
medicare_tax = get_medicare_tax(750)
print(f'Medicare Tax: ${medicare_tax:.2f}')

# Federal Tax
total_tax = get_federal_tax(750, 0)
print(f'Federal Tax: ${total_tax:.2f}')
```

### Person 2

```python
# Social Security Tax
ss_tax = get_soc_sec_tax(1550)
print(f'Social Security Tax: ${ss_tax:.2f}')

# Medicare Tax
medicare_tax = get_medicare_tax(1550)
print(f'Medicare Tax: ${medicare_tax:.2f}')

# Federal Tax
total_tax = get_federal_tax(1550, 2)
print(f'Federal Tax: ${total_tax:.2f}')
```

### Person 3

```python
# Social Security Tax
ss_tax = get_soc_sec_tax(1100)
print(f'Social Security Tax: ${ss_tax:.2f}')

# Medicare Tax
medicare_tax = get_medicare_tax(1100)
print(f'Medicare Tax: ${medicare_tax:.2f}')

# Federal Tax
total_tax = get_federal_tax(1100, 5)
print(f'Federal Tax: ${total_tax:.2f}')
```
