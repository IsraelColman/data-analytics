```python
def favorite_things(name, movie, music):
    return f'Name: {name}\nFavorite Movie: {movie}\nFavorite Musician/Band: {music}'

print(favorite_things("Israel", "Dune Part 2", "Michael Jackson"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here('I am in the Year Up United program because I want to grow my career and network'))
```

```python
def favorite_place(place, why):
    return f'One of my favorite places is {place} Because {why}'

print(favorite_place('Cancun', 'the weather is nice, the food is good, and there are a lot of things to do'))
```

```python
print(favorite_place('My gaming setup', 'I get to relax and take my mind off things while I play games with my friends'))
```
